```python
from safetensors import safe_open

path = "/content/whisper.cpp/zero-stt-hinglish/model.safetensors"

try:
    with safe_open(path, framework="pt", device="cpu") as f:
        keys = list(f.keys())
        print("✅ Valid safetensors file")
        print("Number of tensors:", len(keys))
        print("First 10 keys:", keys[:10])
except Exception as e:
    print("❌ Safetensors error:")
    print(type(e).__name__, e)
```

```python
from huggingface_hub import hf_hub_download
import os

repo_id = "shunyalabs/zero-stt-hinglish"
local_dir = "/content/whisper.cpp/zero-stt-hinglish"

# Remove the 135-byte Git LFS pointer
pointer = os.path.join(local_dir, "model.safetensors")
if os.path.exists(pointer):
    os.remove(pointer)

# Download the REAL model weights
downloaded = hf_hub_download(
    repo_id=repo_id,
    filename="model.safetensors",
    local_dir=local_dir
)

print("Downloaded:", downloaded)
print("Size:", os.path.getsize(downloaded), "bytes")
```
